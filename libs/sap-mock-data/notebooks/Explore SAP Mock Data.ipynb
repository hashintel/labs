# Explore SAP Mock Data
Inspect a warehouse generated by the library or CLI.

```python
import os
from pathlib import Path
import pandas as pd
from sap_mock_data.storage import DeltaTableStore
from sap_mock_data.validation import build_manifest, integrity_report

warehouse = Path(os.environ.get('SAP_MOCK_WAREHOUSE', '.mock-warehouse')).resolve()
store = DeltaTableStore(warehouse)
manifest = build_manifest(store)
pd.DataFrame(
    [(name, details['rows']) for name, details in manifest['tables'].items()],
    columns=['table', 'rows'],
).sort_values('table')
```

```python
integrity_report(store)
```
